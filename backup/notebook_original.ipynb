## 1. Define the Problem

For this project, I am building a machine learning model to predict whether a loan applicant will **default** or **fully repay** their loan. This is a critical task for financial institutions, as minimizing defaults directly reduces financial losses.

* **What to predict:** Whether a borrower will default (`Default = 1`) or fully repay (`Default = 0`).
* **Who will use it:** A loan officer or an automated banking application.
* **ML Problem Type:** Supervised Binary Classification.
* **Success Metrics:** 
  * **Primary:** ROC-AUC (best for evaluating imbalanced datasets).
  * **Secondary:** Recall on the default class (to catch as many bad loans as possible).
  * **Reference:** F1-Score and Confusion Matrix.
